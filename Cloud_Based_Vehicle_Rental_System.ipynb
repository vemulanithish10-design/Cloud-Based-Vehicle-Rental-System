```python
# Cloud-Based Vehicle Rental System

vehicles = {
    "1": ["Car", 1500],
    "2": ["Bike", 500],
    "3": ["Scooter", 300],
    "4": ["SUV", 2500]
}

print("===================================")
print(" CLOUD-BASED VEHICLE RENTAL SYSTEM")
print("===================================")

print("\nAvailable Vehicles:")

for key, value in vehicles.items():
    print(key, "-", value[0], "- ₹", value[1], "per day")

# Customer details
name = input("\nEnter Customer Name: ")

choice = input("Enter Vehicle Number: ")

days = int(input("Enter Number of Days: "))

# Rental calculation
vehicle = vehicles[choice][0]
price = vehicles[choice][1]

total = price * days

print("\n===================================")
print("          RENTAL DETAILS")
print("===================================")
print("Customer Name :", name)
print("Vehicle       :", vehicle)
print("Days          :", days)
print("Price/Day     : ₹", price)
print("Total Cost    : ₹", total)
print("===================================")
print("Vehicle rented successfully!")
```
